# Analyse des résultats obtenus pour l'évaluation utilisateur

```python
import pandas as pd
import krippendorff

df = pd.read_csv("eval_results_realtime.csv")
df['unique_query'] = df['dataset'] + "_" + df['query_id'].astype(str)

#Notes moyennes
print(" NOTES MOYENNES (sur 5) ")
print(f"Baseline : {df['score_baseline'].mean():.2f}")
print(f"T-WSLS : {df['score_twsls'].mean():.2f}\n")

# Win / Tie / Loss
diff = df['score_twsls'] - df['score_baseline']
win = (diff > 0).mean() * 100
tie = (diff == 0).mean() * 100
loss = (diff < 0).mean() * 100
print("COMPARAISON (Win / Tie / Loss)")
print(f"Win (T-WSLS meilleur) : {win:.1f}%")
print(f"Tie (égalité) : {tie:.1f}%")
print(f"Loss(Baseline meilleur): {loss:.1f}%\n")

# Accord inter-annotateurs
def kripp_interval(df, score_col):
    pivot = df.pivot(index='username', columns='unique_query', values=score_col)
    return krippendorff.alpha(reliability_data=pivot.to_numpy(),
                              level_of_measurement='interval')

alpha_b = kripp_interval(df, 'score_baseline')
alpha_t = kripp_interval(df, 'score_twsls')

print("ACCORD INTER-ANNOTATEURS (Krippendorff)")
print(f"Baseline : {alpha_b:.3f}")
print(f"T-WSLS : {alpha_t:.3f}")

```
